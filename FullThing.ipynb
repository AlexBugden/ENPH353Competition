```python
from PIL import Image
import os
import glob

def batch_convert_png_to_jpg(input_dir, output_dir="/content/input", quality=95):
    """
    Convert all PNGs in a directory to JPGs in /content/input

    Args:
        input_dir (str): Directory containing PNG files
        output_dir (str): Output directory (default: /content/input)
        quality (int): JPG quality (1-100)
    """
    os.makedirs(output_dir, exist_ok=True)
    png_files = glob.glob(os.path.join(input_dir, "*.png"))

    if not png_files:
        print("⚠️ No PNG files found in the input directory")
        return

    for png_path in png_files:
        try:
            with Image.open(png_path) as img:
                if img.mode == 'RGBA':
                    white_bg = Image.new('RGB', img.size, (255, 255, 255))
                    white_bg.paste(img, mask=img.split()[3])
                    img = white_bg

                filename = os.path.basename(png_path)
                jpg_path = os.path.join(output_dir, filename.replace('.png', '.jpg'))
                img.save(jpg_path, 'JPEG', quality=quality)
                print(f"Converted {filename}")

        except Exception as e:
            print(f"Failed to convert {png_path}: {str(e)}")

    print(f"\n✅ Successfully converted {len(png_files)} PNG files to JPG in {output_dir}")

# Example usage:
batch_convert_png_to_jpg("/content/processed")
```

```python
import cv2
import numpy as np
import os
from ultralytics import YOLO
import matplotlib.pyplot as plt

# Load trained model
model = YOLO("/content/best.pt")

# Run inference and visualize
image_path = "/content/input/CRIMEPILLOWFIGHT:1.jpg"
results = model(image_path)  # Perform detection

# Load and display original image with bounding boxes
img = cv2.imread(image_path)
img_with_boxes = img.copy()

# Create directory for saving letters
letters_dir = "/content/letters/"
os.makedirs(letters_dir, exist_ok=True)

# Process each detected word
for i, result in enumerate(results):
    for j, box in enumerate(result.boxes.xyxy):
        x1, y1, x2, y2 = map(int, box)

        # Draw bounding box on visualization image
        cv2.rectangle(img_with_boxes, (x1, y1), (x2, y2), (0, 255, 0), 2)

        # Crop the word
        word_crop = img[y1:y2, x1:x2]

        # Convert to grayscale and threshold to find letter separation
        gray = cv2.cvtColor(word_crop, cv2.COLOR_BGR2GRAY)
        _, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

        # Find vertical projection profile
        vertical_projection = np.sum(thresh, axis=0)

        # Find gaps between letters (where projection is zero)
        gap_threshold = 1  # Minimum gap width to consider as letter separation
        in_gap = False
        gap_start = 0
        split_positions = []

        for pos, value in enumerate(vertical_projection):
            if value == 0 and not in_gap:
                in_gap = True
                gap_start = pos
            elif value > 0 and in_gap:
                in_gap = False
                if pos - gap_start > gap_threshold:
                    split_positions.append((gap_start + pos) // 2)  # Middle of the gap

        # Add start and end positions
        split_positions = [0] + split_positions + [word_crop.shape[1]]

        # Split word into letters using found positions
        for k in range(len(split_positions)-1):
            letter_x1 = split_positions[k]
            letter_x2 = split_positions[k+1]

            # Skip if too narrow (likely not a real letter)
            if letter_x2 - letter_x1 < 5:
                continue

            letter_crop = word_crop[:, letter_x1:letter_x2]

            # Save each letter
            letter_filename = os.path.join(letters_dir, f"letter_{i}_{j}_{k}.png")
            cv2.imwrite(letter_filename, letter_crop)

            # Draw letter boxes for visualization
            cv2.rectangle(img_with_boxes,
                         (x1 + letter_x1, y1),
                         (x1 + letter_x2, y2),
                         (255, 0, 0), 1)

# Display results
plt.figure(figsize=(15, 10))
plt.imshow(cv2.cvtColor(img_with_boxes, cv2.COLOR_BGR2RGB))
plt.title("Word Detection (Green) and Letter Segmentation (Blue)")
plt.axis('off')
plt.show()

print(f"✅ Letter images saved to {letters_dir}")
```

```python
import cv2
import numpy as np
from PIL import Image
from ultralytics import YOLO
from tensorflow.keras.models import load_model
import io

# Class to letter mapping
class_to_letter = {
    0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E',
    5: 'F', 6: 'G', 7: 'H', 8: 'I', 9: 'J',
    10: 'K', 11: 'L', 12: 'M', 13: 'N', 14: 'O',
    15: 'P', 16: 'R', 17: 'S', 18: 'T', 19: 'U',
    20: 'V', 21: 'W', 22: 'Y', 23: 'Z'
}

def prepare_image_for_prediction(img_array):
    """Prepare image array for prediction (same as original but takes array instead of path)"""
    # Resize with cv2 (same as original)
    img_resized = cv2.resize(img_array, (183, 256))  # Note: (width, height) is reversed from resize dimensions

    # Normalize (same as original)
    img_normalized = img_resized / 255.0

    # Add batch dimension
    img_final = np.expand_dims(img_normalized, axis=0)

    return img_final

def order_points(pts):
    """Orders the points in consistent order (unchanged from original)"""
    rect = np.zeros((4, 2), dtype="float32")
    s = pts.sum(axis=1)
    rect[0] = pts[np.argmin(s)]
    rect[2] = pts[np.argmax(s)]
    diff = np.diff(pts, axis=1)
    rect[1] = pts[np.argmin(diff)]
    rect[3] = pts[np.argmax(diff)]
    return rect
def process_image(image, min_box_area=80000, debug=False):
    """Process image to find and transform largest blue box if it meets minimum size requirement"""
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    lower_blue = np.array([102, 124, 80])
    upper_blue = np.array([140, 255, 255])
    mask = cv2.inRange(hsv, lower_blue, upper_blue)

    if debug:
        cv2_imshow(mask)
        cv2.waitKey(0)

    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    if not contours:
        print("No blue box detected")
        return None

    # Find largest contour
    largest_contour = max(contours, key=cv2.contourArea)
    contour_area = cv2.contourArea(largest_contour)

    if debug:
        # Draw the largest contour on a copy of the image
        debug_img = image.copy()
        cv2.drawContours(debug_img, [largest_contour], -1, (0, 255, 0), 3)
        cv2.putText(debug_img, f"Area: {contour_area:.0f}", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
        cv2_imshow(debug_img)
        cv2.waitKey(0)

    if contour_area < min_box_area:
        print(f"Blue box too small (area: {contour_area} < minimum: {min_box_area})")
        return None

    epsilon = 0.02 * cv2.arcLength(largest_contour, True)
    approx = cv2.approxPolyDP(largest_contour, epsilon, True)

    if len(approx) == 4:
        pts_src = np.array([point[0] for point in approx], dtype="float32")
        ordered_pts = order_points(pts_src)
        width, height = 300, 150
        pts_dst = np.array([
            [0, 0], [width - 1, 0],
            [width - 1, height - 1], [0, height - 1]
        ], dtype="float32")
        M = cv2.getPerspectiveTransform(ordered_pts, pts_dst)
        warped = cv2.warpPerspective(image, M, (width, height))
        return warped

    print("Could not find a clear rectangle")
    return None

def remove_blue_border(image):
    """Remove blue border from image (now takes/returns image instead of paths)"""
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    lower_blue = np.array([0, 2, 0])
    upper_blue = np.array([255, 255, 255])
    mask = cv2.inRange(hsv, lower_blue, upper_blue)
    mask_inv = cv2.bitwise_not(mask)
    kernel = np.ones((3,3), np.uint8)
    mask_inv = cv2.morphologyEx(mask_inv, cv2.MORPH_CLOSE, kernel, iterations=2)
    contours, _ = cv2.findContours(mask_inv, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    if contours:
        x, y, w, h = cv2.boundingRect(np.concatenate(contours))
        cropped = image[y:y+h, x:x+w]
        return cropped
    return image

def convert_to_jpg(image):
    """Convert image to JPG format in memory (replaces PNG to JPG file conversion)"""
    # Convert color space if needed
    if len(image.shape) == 2:  # Grayscale
        image = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)
    elif image.shape[2] == 4:  # RGBA
        image = cv2.cvtColor(image, cv2.COLOR_RGBA2BGR)

    # Encode as JPG in memory
    _, jpg_image = cv2.imencode('.jpg', image)
    return cv2.imdecode(jpg_image, cv2.IMREAD_COLOR)

def process_words(image, yolo_model):
    """Process image to detect words and return letters grouped by words"""
    results = yolo_model(image)
    img_with_boxes = image.copy()
    word_groups = []  # This will store lists of letters for each word

    for i, result in enumerate(results):
        # Sort boxes by x-coordinate to maintain left-to-right order
        sorted_boxes = sorted(result.boxes.xyxy, key=lambda box: box[0])

        for j, box in enumerate(sorted_boxes):
            x1, y1, x2, y2 = map(int, box)
            cv2.rectangle(img_with_boxes, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Crop the word
            word_crop = image[y1:y2, x1:x2]
            word_letters = []  # Letters for this specific word

            # Convert to grayscale and threshold to find letter separation
            gray = cv2.cvtColor(word_crop, cv2.COLOR_BGR2GRAY)
            _, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

            # Find vertical projection profile
            vertical_projection = np.sum(thresh, axis=0)

            # Find gaps between letters
            gap_threshold = 1
            in_gap = False
            gap_start = 0
            split_positions = []

            for pos, value in enumerate(vertical_projection):
                if value == 0 and not in_gap:
                    in_gap = True
                    gap_start = pos
                elif value > 0 and in_gap:
                    in_gap = False
                    if pos - gap_start > gap_threshold:
                        split_positions.append((gap_start + pos) // 2)

            split_positions = [0] + split_positions + [word_crop.shape[1]]

            # Split word into letters
            for k in range(len(split_positions)-1):
                letter_x1 = split_positions[k]
                letter_x2 = split_positions[k+1]

                if letter_x2 - letter_x1 >= 5:
                    letter_crop = word_crop[:, letter_x1:letter_x2]
                    word_letters.append(letter_crop)

                    # Draw letter boxes for visualization
                    cv2.rectangle(img_with_boxes,
                                 (x1 + letter_x1, y1),
                                 (x1 + letter_x2, y2),
                                 (255, 0, 0), 1)

            word_groups.append(word_letters)

    return word_groups, img_with_boxes

def recognize_letters(word_groups, char_model):
    """Recognize letters while maintaining word groupings"""
    recognized_words = []

    for word_letters in word_groups:
        word_text = []
        for letter_img in word_letters:
            # Convert to PIL Image format
            pil_img = Image.fromarray(cv2.cvtColor(letter_img, cv2.COLOR_BGR2RGB))
            img_array = np.array(pil_img)

            # Prepare for prediction
            prepared = prepare_image_for_prediction(img_array)

            if prepared is not None:
                prediction = char_model.predict(prepared)
                predicted_class = np.argmax(prediction)
                word_text.append(class_to_letter.get(predicted_class, '?'))

        recognized_words.append(''.join(word_text))

    return recognized_words



def process_complete_image(image_path, yolo_model, char_model):
    """Complete processing pipeline for a single image"""
    # Load initial image (only file operation remaining)
    original_image = cv2.imread(image_path)
    if original_image is None:
        print(f"Failed to load image: {image_path}")
        return []

    # Step 1: Process image to find and transform blue box
    transformed = process_image(original_image)
    if transformed is None:
        return []

    # Step 2: Remove blue border
    cleaned = remove_blue_border(transformed)

    # Step 3: Convert to JPG (in memory)
    jpg_image = convert_to_jpg(cleaned)

    # Step 4: Detect words and get letter groups
    word_groups, visualization = process_words(jpg_image, yolo_model)

    # Step 5: Recognize letters while maintaining word structure
    recognized_words = recognize_letters(word_groups, char_model)

    return recognized_words, visualization

# Define our special words and their corresponding IDs
SPECIAL_WORDS = {
    'SIZE': 1,
    'VICTIM': 2,
    'CRIME': 3,
    'TIME': 4,
    'PLACE': 5,
    'MOTIVE': 6,
    'WEAPON': 7,
    'BANDIT': 8
}

def find_special_word(word, special_words):
    """Check if word matches any special word with 1 character substitution allowed"""
    word = word.upper()  # Convert to uppercase for comparison

    for special in special_words:
        # Exact match
        if word == special:
            return special

        # Only check words of same length
        if len(word) == len(special):
            # Count character differences
            diff_count = sum(1 for a, b in zip(word, special) if a != b)
            if diff_count <= 1:  # Allow 1 character substitution
                return special
    return None

def process_recognized_words(recognized_words):
    """Process the recognized words to find special words and create message"""
    id = None
    message_words = []

    for word in recognized_words:
        special_word = find_special_word(word, SPECIAL_WORDS.keys())
        if special_word:
            if id is None:  # Only take the first special word found
                id = SPECIAL_WORDS[special_word]
        else:
            message_words.append(word)

    message = ' '.join(message_words) if message_words else None
    return id, message


```
